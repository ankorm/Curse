### Пример из текста методички

Цикл `while` с условием `a.all()`: элементы уменьшаются, пока хотя бы один из них не станет равным нулю. С условием `a.any()` тот же цикл без `break` не завершится: элементы обнуляются в разные моменты.

```python
a = np.array([2, 5, 3, 4, 5])
while a.all():
    a = a - 1
print(a)    # [0 3 1 2 3]
```

### Пример из текста методички

Переключатель `match`–`case` (Python 3.10+).

```python
for var in [3, 'May', 10, 'Autumn']:
    match var:
        case 1 | 2 | 12 | 'January' | 'February' | 'December':
            print(var, '->', 'Winter')
        case 3 | 4 | 5 | 'March' | 'April' | 'May':
            print(var, '->', 'Spring')
        case 6 | 7 | 8 | 'June' | 'July' | 'August':
            print(var, '->', 'Summer')
        case 9 | 10 | 11 | 'September' | 'October' | 'November':
            print(var, '->', 'Autumn')
        case _:
            print(var, '->', 'It is not a month.')
```
